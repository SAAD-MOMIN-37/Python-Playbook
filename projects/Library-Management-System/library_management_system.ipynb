# 📚 Library Management System
A menu-driven CRUD application built with Python and JSON.

## 🎯 Objective
Implement CRUD operations, modular programming, and JSON persistence.

## 🛠️ Technologies
- Python
- JSON
- Jupyter Notebook

## 🚀 Features
- Add Book
- Display Books
- Search Book
- Update Book
- Delete Book
- Save/Load JSON

```python
books = [
    {
        "book_id": 101,
        "title": "Python Programming",
        "author": "Guido van Rossum",
        "category": "Programming",
        "copies": 5
    },
    {
        "book_id": 102,
        "title": "Machine Learning",
        "author": "Tom Mitchell",
        "category": "AI",
        "copies": 3
    },
    {
        "book_id": 103,
        "title": "Deep Learning",
        "author": "Ian Goodfellow",
        "category": "AI",
        "copies": 4
    }
]
```

```python
books
```

```python
def add_book():
    book_id = int(input("Enter book id:"))
    title = input("Enter book title:")
    author = input("Enter author name:")
    category = input("Enter book category:")
    copies = int(input("Enter number of copies:"))

    new_book = {
        'book_id': book_id,
        'title': title,
        'author': author,
        'category': category,
        'copies': copies
    }

    for book in books:
       if new_book["book_id"] == book["book_id"]:
           book['copies'] += copies
           print("-"*70)
           print("✅Book is updated!!")
           print("-"*70)
           return
       
    books.append(new_book)
    print("-"*70)
    print("✅Book is added!!")
    print("-"*70)
    return

add_book()
```

```python
books
```

```python
def book_info(book):
    print("-"*70)
    print(f"\nBook ID       : {book["book_id"]}")
    print(f"Title         : {book["title"]}")
    print(f"Author        : {book["author"]}")
    print(f"Category      : {book["category"]}")
    print(f"No. of Copies : {book["copies"]}\n")
    print("-"*70)
    return

book_info(books[0])
```

```python
def display_book():
        if not books:
            print("No books found!!")
            return
        for book in books:
            book_info(book)
        

display_book()
```

```python
def search_book():
    search = int(input("Enter Book ID to search:"))
    if not books:
        print("No book found!!")
        return 
    
    for book in books:
        if search == book["book_id"]:
            book_info(book)
            return
            
    print("-"*70)
    print("Book ID not found!!")
    print("-"*70) 
    return
            
search_book()
```

```python
def update_book():
    update = int(input("Enter Book ID to update:"))
    if not books:
        print("No book found!!")
        return
    for book in books:
        if update == book["book_id"]:
            print("-"*70)
            print("Before Updating...")
            book_info(book)
            book["title"] = input("Enter Book Title:")
            book["author"] = input("Enter Book Author:")
            book["category"] = input("Enter Book Category:") 
            book["copies"] = int(input("Enter No. of Copies:"))
            print("-"*70)
            print("After Updating...")
            book_info(book)
            return
    print("-"*70)
    print("Book ID not found!!")
    print("-"*70)
    return

update_book()
```

```python
def delete_book():
    delete = int(input("Enter Book ID ")) 
    if not books:
        print("No book found!!")
        return
    for book in books:
        if delete ==  book["book_id"]:
            print("Before Deleting...")
            book_info(book)

            choice = input("Are you sure you want to delete this book (y/n)").lower()
            if choice in ['y','yes']:
                books.remove(book)
                print("-"*70)
                print("✔️ Book deleted successfully!")
                print("-"*70)
                return
            else:
                print("-"*70)
                print("❌ Deletion Cancelled!!")
                print("-"*70)
                return
    print("-"*70)
    print("Book ID not found")
    print("-"*70)
    return

delete_book()
```

```python
import json
def save_data(file = "books.json"):
    with open(file, "w") as f:
        json.dump(books, f,indent = 4)
    print("File saved successfully!")

save_data()
```

```python
def load_data(file = "books.json"):
    global books
    with open(file, "r") as f:
        books = json.load(f)
    print("File loaded Successfully!")
load_data()
```

```python
def system():

    load_data()

    while True:

        print("=" * 70)
        print("                 LIBRARY MANAGEMENT SYSTEM")
        print("=" * 70)

        print("\n1. Add Book")
        print("2. Display Books")
        print("3. Search Book")
        print("4. Update Book")
        print("5. Delete Book")
        print("6. Save Data")
        print("7. Exit\n")

        print("=" * 70)

        try:

            choice = int(input("Enter your choice (1-7): "))

            if choice == 1:
                add_book()

            elif choice == 2:
                display_book()

            elif choice == 3:
                search_book()

            elif choice == 4:
                update_book()

            elif choice == 5:
                delete_book()

            elif choice == 6:
                save_data()

            elif choice == 7:
                save_data()
                print("-" * 70)
                print("💖 Thank you for using Library Management System!")
                print("-" * 70)
                break

            else:
                print("❌ Please enter a number between 1 and 7.")

        except ValueError:
            print("-" * 70)
            print("❌ Invalid input! Please enter a valid number.")
            print("-" * 70)
system()
```
