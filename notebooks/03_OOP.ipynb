**Q4**

Create a `BankAccount` class using **encapsulation** (a private balance attribute) with `deposit()`, `withdraw()`, and `get_balance()` methods.

```python
class BankAccount:

    def __init__(self,account_holder,balance):
        self.account_holder = account_holder
        self.__balance = balance

    def get_balance(self):
        return self.__balance
        
    def deposit(self,amount):
        self.__balance += amount
        return f"{amount} Added!"

    def withdraw(self,amount):
        if (self.__balance >= amount) and (self.__balance > 0) :
            self.__balance -= amount
            return f"{amount} Withdrawed!"
        else:
            print(f"Your account does not have such amount balance!!") 
            print(f"please deposit some amount to withdraw!")


b1 = BankAccount("Rahul",0)
b1.get_balance()
b1.deposit(5000)
b1.withdraw(5000)
b1.get_balance()
```

**Q6**

Create a `Laptop` class using a **class method** to change the class-level `brand` attribute.

```python
class Laptop:
    brand = "HP"
    def __init__(self,name):
        self.name = name
        
    def display_info(self):
        print(f"Name : {self.name}") 
        print(f"Brand : {self.brand}") 

    @classmethod
    def change_brand(cls, brand):
        cls.brand = brand

    def change_name(self,name):
        self.name = name 
        
l1 = Laptop("Victus")
l1.display_info()
l1.change_brand("Lenovo")
l1.change_name("LOQ")
l1.display_info()
```

**Q9**

Use **multilevel inheritance** to create `Person` → `Employee` → `Manager` classes.

```python
class Person:

    def __init__(self,name,age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name : {self.name}")
        print(f"Age : {self.age}")

class Employee(Person):

    def __init__(self,name,age,salary,emp_id):
        super().__init__(name,age)
        self.salary = salary
        self.emp_id = emp_id

    def display(self):
        super().display()
        print(f"Salary : {self.salary}")
        print(f"Employee ID : {self.emp_id}")

class Manager(Employee):

    def __init__(self,name,age,salary,emp_id):
        super().__init__(name,age,salary,emp_id)

    def display(self):
        super().display()

m = Manager("Ankit",35,60000,201)
m.display()
```

**Q10**

Use **multiple inheritance** to create a `Child` class that inherits from both `Father` and `Mother` classes.

```python
class Father:

    def __init__(self, fname, focc):
        self.fname = fname
        self.focc = focc

    def display_father(self):
        print("Father Information")
        print(f"Name       : {self.fname}")
        print(f"Occupation : {self.focc}")


class Mother:

    def __init__(self, mname, mocc):
        self.mname = mname
        self.mocc = mocc

    def display_mother(self):
        print("Mother Information")
        print(f"Name       : {self.mname}")
        print(f"Occupation : {self.mocc}")


class Child(Father, Mother):

    def __init__(self, fname, focc, mname, mocc, cname, age):

        Father.__init__(self, fname, focc)
        Mother.__init__(self, mname, mocc)

        self.cname = cname
        self.age = age

    def display_child(self):
        print("Child Information")
        print(f"Name : {self.cname}")
        print(f"Age  : {self.age}")


# Input
father_name = input("Enter Father's Name: ")
father_occ = input("Enter Father's Occupation: ")

mother_name = input("\nEnter Mother's Name: ")
mother_occ = input("Enter Mother's Occupation: ")

child_name = input("\nEnter Child's Name: ")
child_age = int(input("Enter Child's Age: "))

# Object
c = Child(
    father_name,
    father_occ,
    mother_name,
    mother_occ,
    child_name,
    child_age
)

print("\n" + "-" * 50)
print("FAMILY DETAILS")
print("-" * 50)

c.display_father()
print()

c.display_mother()
print()

c.display_child()
```

**Q12**

Use **abstraction** with the `abc` module to create an abstract `Shape` class implemented by `Circle` and `Rectangle`.

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

c = Circle(5)
r = Rectangle(10, 4)

print(f"Area of Circle    : {c.area():.2f}")
print(f"Area of Rectangle : {r.area()}")
```

**Q13**

Use **magic methods** (`__str__` and `__len__`) in a `Book` class to customize its string representation and length.

```python
class Book:

    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f"Book: {self.title}\nAuthor: {self.author}"

    def __len__(self):
        return self.pages


book = Book("Python Programming", "Saad Momin", 350)

print(book)
print(f"Total Pages : {len(book)}")
```
