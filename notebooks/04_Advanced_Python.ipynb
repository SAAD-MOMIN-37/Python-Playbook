## 📚 Section 1 — Comprehensions

**Q1**

Use a list comprehension to generate the squares of numbers from 1 to 20.

```python
square = [x**2 for x in range(1,21)]
print(square)
```

**Q2**

Use a list comprehension with a conditional filter to generate all even numbers between 1 and 50.

```python
even = [x for x in range(1,51) if (x % 2) == 0 ]
print(even)
```

**Q3**

Use a dictionary comprehension to map numbers 1 to 10 to their squares.

```python
sq = {x:x**2 for x in range(1,11)}
sq
```

**Q4**

Use a set comprehension to generate the cubes of numbers 1 to 10.

```python
cube = set(x**3 for x in range(1,11))
cube
```

**Q5**

Use a nested list comprehension to create a 3x3 matrix initialized with zeros.

```python
matrix = [[0 for j in range(3)] for i in range(3)]
matrix
```

## 📚 Section 2 — Iterators & Generators

**Q6**

Create an iterator from a list using `iter()` and retrieve elements one at a time using `next()`.

```python
fruits = ["Apple","Mango","Banana","Pineapple","Orange"]
iterate = iter(fruits)
print(next(iterate))
print(next(iterate))
print(next(iterate))
print(next(iterate))
print(next(iterate))
```

**Q7**

Use an iterator with a try/except block to manually iterate over a tuple until `StopIteration` is raised.

```python
num = (1,2,3,4,5)
it = iter(num)
try:
    while True:
        print(next(it))
except StopIteration:
    print("Iteration Completed!")
```

**Q8**

Create a generator function that yields numbers from 1 to `n`.

```python
def generator(n):
    for i in range(1,n+1):
        yield i

gen = generator(10)
for i in gen:
    print(i)
```

**Q9**

Create a generator function that yields even numbers up to 50.

```python
def even_geneeator():
    for i in range(2,51,2):
        yield i

gen = even_geneeator()
for i in gen:
    print(i)
```

**Q10**

Use a generator expression to yield the squares of numbers from 1 to 10.

```python
generator = (x**2 for x in range(1,11))
for i in generator:
    print(i)
```

## 📚 Section 3 — Advanced Functions

**Q11**

Assign a function to a variable and call it through that variable to demonstrate that functions are first-class objects.

```python
def greet():
    return "Welcome to Python!"
msg = greet
msg()
```

**Q12**

Pass a function as an argument to another function and invoke it from inside.

```python
def square():
    x = int(input("Enter a number: "))
    print(x**2)

def display(func):
    func()

display(square)
```

**Q13**

Create a closure where an inner function captures and uses a variable from its enclosing scope.

```python
def outer():

    x = "Learning Advanced Python"

    def inner():
        print(x)

    return inner

fun = outer()
fun()
```

**Q14**

Create a closure-based function factory that returns a customized multiplier function.

```python
def multiplier(x):
    def another(n):
        return x * n
    return another

double = multiplier(2)
print(double(5))
```

**Q15**

Create a decorator that adds behavior before and after the wrapped function's call.

```python
def decorator(func):
    def wrapper():
        print("Program Started")
        func()
        print("Program Ended")
    return wrapper

@decorator
def welcome():
    print("Welcom!!")

welcome()
```

Use `enumerate()` to loop over a list while tracking both index and value.

**Q18**

Use `zip()` to iterate over two lists in parallel.

```python
names = ["Saad","Bob","Rohan","Ali","Lily"]
marks = [56,59,69,86,78]

for name, mark in zip(names,marks):
    print(name,"-->",mark)
```

**Q19**

Use `any()` with a list comprehension to check if at least one value satisfies a condition.

```python
nums = [23, 56, 34]
passed =[num>=35 for num in nums]
print(any(passed))
```

**Q20**

Use `all()` with a list comprehension to check if every value satisfies a condition.

```python
nums = [-96, 56, 23]
pos = [num>0 for num in nums]
print(all(pos))
```
