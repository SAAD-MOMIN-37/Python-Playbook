**Q13**

Find the largest among three numbers.



```python
num1 = int(input("Enter First Number:")) 
num2 = int(input("Enter Second Number:")) 
num3 = int(input("Enter Third Number:")) 

if num1 > num2 and num1 > num3:
    print(f"The First number is largest.")
elif num2 > num1 and num2 > num3:
    print(f"The Second number is largest.") 
elif num3 > num1 and num3 > num2:
    print(f"The Third number is largest.") 
else:
    print("two or more numbers are equal")
```

**Q15**

Print the multiplication table of a number using a `while` loop.

Input:

Number


```python
num = int(input("Enter a Number: "))
i = 1

print(f"\nMultiplication Table of {num}\n")

while i <= 10:
    print(f"{num} x {i} = {num * i}")
    i += 1
```
